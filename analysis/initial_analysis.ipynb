```python
for problem in DB["problem"].unique():
      print(f"Problem: {problem}")
      subset = DB[DB["problem"] == problem]
      for model, group in subset.groupby("model", dropna=False):
          n_params = group["n_params"].dropna()
          if n_params.empty:
              print(f"  Model: {model}, n_params: missing")
          else:
              print(f"  Model: {model}, n_params: {n_params.iloc[0]}")
```

```python
for problem in DB["problem"].unique():
    print(f"Problem: {problem}")
    subset = DB[DB["problem"] == problem]
    for model, group in subset.groupby("model", dropna=False):
        n_params_vals = sorted(group["n_params"].dropna().unique())
        if not n_params_vals:
            print(f"  Model: {model}, n_params: missing")
        else:
            print(f"  Model: {model}, n_params: {n_params_vals}")
```
